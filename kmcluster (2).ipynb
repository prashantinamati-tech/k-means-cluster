## k-means cluster
#### identify k cluster(grouping) within the datset

##### steps :
> 1)Choose 3(based on k value)  random points as  Centroids(mean)

> 2)Calculate distance bw points and centroids and than assign points to cloest centroid ..

>3)compute new centroids ..

note :  we need to repeat step 2 and step 3 until we reach stability (centroids value gets repeated)

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
iris.data
```

```python
iris.feature_names
```

```python
columns = iris.feature_names
```

```python
columns
```

```python
df = pd.DataFrame(iris.data, columns=columns)
```

```python
df
```

```python
### splitting as training and testing
from sklearn.model_selection import train_test_split
train ,test = train_test_split(df, test_size=0.2, random_state=42)
```

```python
train.shape
```

```python
test.shape
```

```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=3,random_state=42)
```

```python
model.fit(train)
```

```python
len(model.labels_)
```

```python
predict = model.predict(test)
```

```python
predict
```

```python
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
sns.scatterplot(x=test['petal length (cm)'], y=test['petal width (cm)'], hue=predict)
```
